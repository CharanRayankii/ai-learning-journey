```python
corpus="""Hello Welcome,to Krish Naik's NLP Tutorials.
Please do watch the entire course! to become expert in NLP.
"""
```

```python
print(corpus)
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
##  Tokenization
## Sentence-->paragraphs
from nltk.tokenize import sent_tokenize
```

```python
#converting paragraph into sentences 
# this will give list of sentences
documents = sent_tokenize(corpus)
```

```python
#this will let us know the type of documents
type(documents)
```

```python
#loop through the sentences
for sentence in documents:
    print(sentence)
```

```python
## Tokenization 
## Paragraph-->words
## sentence--->words
from nltk.tokenize import word_tokenize
```

```python
#paragraph to words
word_tokenize(corpus)
```

```python
#sentence to words
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
#puncutation also convert into words
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
```

```python

#in the above "  ' " also considered as word but this one will combine with the next word
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer=TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```
